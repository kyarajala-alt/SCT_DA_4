```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("marketing_campaign.csv", sep="\t")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df = df.dropna()
```

```python
df = df.drop_duplicates()
```

```python
df['Response'].value_counts().plot(kind='bar')

plt.title("Campaign Responses")
plt.xlabel("Response")
plt.ylabel("Count")

plt.show()
```

```python
df['Income'].plot(kind='hist')

plt.title("Income Distribution")
plt.xlabel("Income")

plt.show()
```

```python
channels = ['NumWebPurchases', 'NumCatalogPurchases', 'NumStorePurchases']

df[channels].sum().plot(kind='bar')

plt.title("Purchases by Marketing Channel")
plt.xlabel("Channel")
plt.ylabel("Total Purchases")

plt.show()
```

```python
df['NumWebVisitsMonth'].plot(kind='hist')

plt.title("Monthly Website Visits")
plt.xlabel("Number of Visits")

plt.show()
```

```python
print("Business Insights:")
print("- Customers prefer web and store purchases more than catalog purchases.")
print("- Higher income customers tend to spend more on products.")
print("- Only a smaller group of customers accepted the campaign.")
print("- Website traffic is high, indicating strong online engagement.")
print("- Web marketing channels may provide better ROI.")
```
